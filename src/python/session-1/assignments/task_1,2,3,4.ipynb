```python
''' Suppose you have this email address “Amit_ml@gmail.edu” 
 - Input Validation: Check if the input string contains exactly one "@" symbol and 
    at least one "." after the "@" symbol. If it's not a valid email, return "Invalid 
    email". 
        
    - Extract Username: Extract and return the part of the email before the "@" 
    symbol. 
        
    - Extract Domain: Extract and return the domain (the part between "@" and the 
    last "."). 
        
    - Check for Domain Ending: Check if the email ends with ".com". If it does, 
    return "Commercial Domain". If it ends with ".edu", return "Educational Domain". 
    Otherwise, return "Other Domain".
'''
email= "Amit_ml@gmail.edu"
#Input Validation 
if email.count('@')==1 and email.count('.')>=1 and email.index('.')> email.index('@') :
    Username=email[0:email.index('@'):1]
    Domain=email[email.index('@')+1:email.index('.'):1]
    #check Domain
    if email[email.index('.'):len(email) :1 ]==".com" :
        DomainEnding="commercial Domain"
    elif  email[email.index('.'):len(email) :1 ]==".edu" :
        DomainEnding="Educational Domain"
    else:
        DomainEnding="Other Domain"
else: 
    print("Invalid email")

print(f" Username: {Username} ")
print(f" Domain: {Domain} ")
print(f" Domain Ending: {DomainEnding}")
```

```python
'''
Encoded Message: ###!!@mocleW EPGTQ!!!6789 
Steps to Decode: 
1. Extract the core part of the message: "mocleW EPGTQ". 
2. Reverse the first word: "mocleW" becomes "Welcome". 
3. Replace shifted vowels in the second word: 
o "EPGTQ": No vowels to change. 
4. Final decoded message: "Welcome PGTQ".
'''
Encoded_Message = "###!!@mocleW EPGTQ!!!6789"
#remove non-core chars
Encoded_Message = Encoded_Message.removeprefix("###!!@")
Encoded_Message = Encoded_Message.removesuffix("!!!6789")

#convert string to a list
Encoded_Message = Encoded_Message.split(" ")
#reverse first word
Encoded_Message[0] = Encoded_Message[0][::-1]
# make E in EPGTQ small  and store it in a char varaiable
transferred_char = Encoded_Message[1][0].lower()
#assign the second word to be PGTQ without the e
Encoded_Message[1] = Encoded_Message[1][1:]
#put the e in the end of Welcom
Encoded_Message[0] = Encoded_Message[0] + transferred_char
#join to make it string again
Decoded_Message = " ".join(Encoded_Message)


print(Decoded_Message)
```

# Task_3

```python
'''
Encoded Message: &&&**$gnirtS PLIO!!@1234 
Steps to Decode: 
1. Extract the core part of the message: "gnirtS PLIO". 
2. Reverse the first word: "gnirtS" becomes "String". 
3. Replace shifted vowels in the second word: 
o "PLIO": Replace I->E and O->U to get "PLEU". 
4. Final decoded message: "String PLEU". 
'''
Encoded_Message = "&&&**$gnirtS PLIO!!@1234"
# remove non-core chars
Encoded_Message= Encoded_Message.removeprefix("&&&**$").removesuffix("!!@1234")
# convert string to a list
Encoded_Message = Encoded_Message.split(" ")
# reverse the first word by slicing
Encoded_Message[0]=Encoded_Message[0][::-1]
# Replace I->E and O->U to get "PLEU" 
Encoded_Message[1]=Encoded_Message[1].replace("I","E",-1).replace("O","U",-1)
#convert back to a string by joining the list
decoded_Message=" ".join(Encoded_Message)

print(decoded_Message)
```

```python
'''
Encoded Message: ##$$$@!yalpstcejorp EPUVT****9887 
Steps to Decode: 
1. Extract the core part of the message: "yalpstcejorp EPUVT". 
2. Reverse the first word: "yalpstcejorp" becomes "projectplay". 
3. Replace shifted vowels in the second word: 
o "EPUVT": Replace E->A, U->O to get "APOVT". 
4. Final decoded message: "projectplay APTOV".
'''
import string


Encoded_Message = "##$$$@!yalpstcejorp EPUVT****9887"
# remove non-core chars
non = string.punctuation + string.digits
Encoded_Message = Encoded_Message.strip(non)
# convert string to a list
Encoded_Message = Encoded_Message.split(" ")
# reverse the first word by slicing
Encoded_Message[0] = Encoded_Message[0][::-1]
# Replace E->A and U->O to get "APOVT"
Encoded_Message[1] = Encoded_Message[1].replace("E", "A",).replace("U", "O")
# convert back to a string by joining the list
decoded_Message = " ".join(Encoded_Message)

print(decoded_Message)
```
