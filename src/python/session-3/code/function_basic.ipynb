```python
def train_model(model_name, *datasets, **hyperparameters):
    '''
    Trains a machine learning model using a variable number of datasets and hyperparameters.

    Parameters:
    -----------
    model_name : str
        The name or identifier of the machine learning model (e.g., "Random Forest").
        
    *datasets : tuple
        Variable length positional arguments (*args). 
        Accepts any number of dataset names, file paths, or data objects.
        
    **hyperparameters : dict
        Variable length keyword arguments (**kwargs). 
        Accepts key-value pairs representing hyperparameter settings 
        (e.g., learning_rate=0.01, n_estimators=100).

    Returns:
    --------
    None
        Prints training status, dataset list, and hyperparameters to the console.

    Example:
    --------
    >>> train_model("SVM", "train.csv", "test.csv", C=1.0, kernel="rbf")
    '''
    
    print(f"Training model: {model_name}")
    
    print("\nDatasets:")
    for dataset in datasets:
        print("_", dataset)
        
    print("\nHyperparameters:")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")


```

```python
# مثال لاستدعاء الدالة:
train_model(
    "Random Forest", 
    "train_data.csv", "val_data.csv", "test_data.csv", 
    n_estimators=100, max_depth=10, learning_rate=0.01
)
```
