```python
import math
''' 
this program takes a number and prints the twin primes for it using a for loop
at the end of the program it prints all twin primes between 3 and 1000

'''
print("Twin Primes:")
for i in range(3, 1000, 2):
    
    for x in range(round(math.sqrt(i))+1):
        first_pr=False
        if x==0 or x==1:
            continue
        if i%x==0:
            first_pr=True
            break
    for x in range(round(math.sqrt(i+2))+1):
        second_prime=False
        if x==0 or x==1:
            continue
        if (i+2)%x==0:
            second_prime=True
            break
    if first_pr==False and second_prime==False:
        print(f"{i} and {i+2}")

    
        
```

### Task 4

```python
def decToBin(num:int=2)->str:
    ''' 
    this function takes a number and prints the binary value for it using a while loop 

    parameters:
            takes an integar parameter with an intial value of 2 
    return:
            print a strings value of the binary value 
    '''
    bin_num=""
    while num>0:
        bin_num+=str(num%2)
        num=num//2
    print(bin_num[::-1])
    
decToBin(36)
```

### Task 5

```python
per = []

for num in range(1, 100):
    divisors = []
    
    for i in range(1, num):
        if num % i == 0:
            divisors.append(i)
            
    if sum(divisors) == num:
        per.append(num)

print(f"perfect numbers: {per}")
```
