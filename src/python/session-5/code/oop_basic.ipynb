## Student class

````python
'''  
this  class is used to create student object with name and age attributes and display method to show the details of the student.
with default values for name and age if not provided during object creation.
attributes:
    name: str, default "Unknown"
    age: int, default 0 
   
    methods:
   
        display: returns a string with the name and age of the student. 

'''

class student:
    def __init__(self, name="Unknown", age=0):
        self.name = name
        self.age = age

    def display(self):
       return f"my Name: {self.name}, my Age: {self.age}"
````

````python
student1=student()
student1.display()
````

````python
'''  
    this class is used to create calculator object with two attributes x and y and methods to perform basic arithmetic operations like addition, subtraction, multiplication and division.
    attributes:
        x: int, default 0
        y: int, default 0
    methods:
        add: returns the sum of x and y
        subtract: returns the difference of x and y
        multiply: returns the product of x and y
        divide: returns the quotient of x and y, if y is not zero, otherwise returns "Cannot divide by zero"


'''

class calculator:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def add(self):
        return self.x + self.y

    def subtract(self):
        return self.x - self.y

    def multiply(self):
        return self.x * self.y

    def divide(self):
        if self.y != 0:
            return self.x / self.y
        else:
            return "Cannot divide by zero"
````

````python
input1 = int(input("Enter first number: "))
input2 = int(input("Enter second number: "))
calc1 = calculator(input1, input2)
print("Addition:", calc1.add())
print("Subtraction:", calc1.subtract())
print("Multiplication:", calc1.multiply())
````

````python
'''  
this program demonstrates the use of abstract base classes (ABC) in Python.
an abstract base class Animal is defined with an abstract method move.
two concrete classes Bird and Cat inherit from Animal and implement the move method.




'''
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    def move(self):
        print("move from bird")

class Cat(Animal):
    def move(self):
        print("move from cat")

# example usage
for a in (Bird(), Cat()):
    a.move()
````
