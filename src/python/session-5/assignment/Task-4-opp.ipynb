```python

class multiplication():
    
    def __init__(self,num):
        self.num=num
        
    def multiplication_table(self)->str:
        ''' 
        this function takes a number and prints the multiplication table for it using a for loop 

        parameters:
                takes an integar parameter with an intial value of 2 
        return:
                print a strings value of the multiplication tabel 
        
        '''
        for i in range(1, 13):
            multi= self.num*i
            statement= f"{self.num} * {i}={multi}"
            print(statement)
    
    
        
        
m1 = multiplication(8)
m1.multiplication_table()
```

```python
import math
''' 
this program takes a number and prints the twin primes for it using a for loop
at the end of the program it prints all twin primes between 3 and 1000

'''
class where_twin_primes:
    def __init__(self,up_range,low_range):
        if low_range%2==0:
            self.low_range=low_range+1
        else:
           self.low_range=low_range
        self.up_range=up_range
        

    def twin_primes(self):
        print("Twin Primes:")
        for i in range(self.low_range, self.up_range, 2):
            
            for x in range(round(math.sqrt(i))+1):
                first_pr=False
                if x==0 or x==1:
                    continue
                if i%x==0:
                    first_pr=True
                    break
            for x in range(round(math.sqrt(i+2))+1):
                second_prime=False
                if x==0 or x==1:
                    continue
                if (i+2)%x==0:
                    second_prime=True
                    break
            if first_pr==False and second_prime==False:
                print(f"{i} and {i+2}")

w_1=where_twin_primes(400,18)
w_1.twin_primes()
```

```python
import math
class prime_factors :

    def __init__(self,num):
        self.num=num
    def what_prime_factors(self):    
        x=self.num
        list1=[]
        while x%2==0:
            list1.append(2)
            x=x/2
        for i in range(3,int(float(self.num)/2+1),2):
            if x%i==0:
                x=x/i
                list1.append(i)
        if x>2:
            list1.append(int(x))
        print(f"the prime factors of {self.num} are {list1}")  

p1= prime_factors(66)
p1.what_prime_factors()
```

```python
class perfect_numbers:
    def __init__(self,low_range,up_range):
        self.low_range=low_range
        self.up_range=up_range
    
    def perfect(self):
        per = []

        for num in range(self.low_range, self.up_range):
            divisors = []
            
            for i in range(1, num):
                if num % i == 0:
                    divisors.append(i)
                    
            if sum(divisors) == num:
                per.append(num)

        print(f"perfect numbers between {self.low_range} and {self.up_range}: {per}")
        
        
        
p1=perfect_numbers(12,500)

p1.perfect()
```
