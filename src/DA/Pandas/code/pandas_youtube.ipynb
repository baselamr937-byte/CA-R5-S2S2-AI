# series

```python
import pandas as pd
```

```python
data= [18, 22, 30, 42, 50]
series_1=pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print(series_1)
series_loc=series_1.loc['a':'c']## loc is used to select data by label
print(series_loc)
apartment=series_1.iloc[0:3]## iloc is used to select data by position
print(apartment)

ss=series_1.loc["a"]
print(ss)
```

```python
series_1.loc["f"]=60## adding a new element to the series
series_1.loc["c"]+=70## adding a value to an existing element in the series

print(series_1)
```

# Data frames

```python
dict_data={'Name': ['Ali', 'Ahmed', 'Mohamed',       'Sara',          'Mona'],
           'Age':   [25,            30,        35,     28,        32],
           'City': ['Cairo', 'Alexandria', 'Giza', 'Cairo', 'Alexandria']}
df=pd.DataFrame(dict_data, index=[2, 3, 4, 5, 6])

## how to add a new column to a dataframe
df['Country']=['Egypt', 'Egypt', 'Egypt', 'Egypt', 'Egypt']

## how to add rows to a dataframe
new_row=pd.DataFrame({'Name': ['Hassan'], 'Age': [40], 'City': ['Cairo'], 'Country': ['Egypt']}, index=[7])
df=pd.concat([df, new_row])
print(df)
```
