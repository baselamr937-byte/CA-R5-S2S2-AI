```python
def convert_dtype(arr,dtype):
    try:
        arr= np.asarray(arr)
        return arr.astype(dtype)
    except TypeError as e :
        raise TypeError(f"Invalid Target data Type:{dtype}")from e
    except ValueError as e :
            raise ValueError(f"Invalid Target data Type:{dtype}")from e
```

```python
lst=([1,2.3,3])
print(lst)
arr=convert_dtype(lst,int)
type(lst[0])
```

```python
def init_weights(n_feature,method="zero",seed=None):
    if not isinstance(n_feature,int):
   
        raise TypeError(f"{n_feature}must be int")
    if n_feature<=0:
        raise TypeError(f"{n_feature} must be >0")
    if method=="random":
        if seed is not None:
            np.random.seed(seed)
        return np.random.randn(n_feature)*0.03
    raise ValueError(f"un supported  method type ")
```


## شرح الكود

هذا الكود يقوم بحفظ مصفوفتين `NumPy` في ملف واحد باسم `file_da.npz` ثم قراءتهما مرة أخرى:



```python
import numpy as np

mat1 = np.ones((3, 4))
mat2 = np.zeros((3, 4))

np.savez('file_da.npz', ones_mat=mat1, zeros_mat=mat2)

with np.load('file_da.npz') as file:
    loaded1 = file['ones_mat']
    loaded2 = file['zeros_mat']
```



### ماذا يحدث؟
- `np.ones((3,4))` → إنشاء مصفوفة 3×4 مليئة بـ 1
- `np.zeros((3,4))` → إنشاء مصفوفة 3×4 مليئة بـ 0
- `np.savez(...)` → حفظ كلا المصفوفتين داخل ملف واحد
- `np.load(...)` → فتح الملف واسترجاع كل مصفوفة حسب اسمها:
  - `file['ones_mat']`
  - `file['zeros_mat']`

### النتيجة التي تم التحقق منها
تم تشغيل نفس المثال في البيئة الحالية، وكانت النتيجة:



```python
(3, 4) (3, 4)
float64 float64
[[1. 1.]
 [1. 1.]]
[[0. 0.]
 [0. 0.]]
```



يعني:
- كل مصفوفة بقيت بنفس الشكل `(3,4)`
- النوع float64 لأن `ones` و `zeros` في `NumPy` يكونان نوع `float64` افتراضيًا

> ملاحظة: `np.savez` يتيح حفظ أكثر من مصفوفة في ملف واحد، بينما `np.save` يخصص لملف واحد فقط.

إذا رغبت، أستطيع أيضًا شرح الفرق بين:
- `np.save`
- `np.savez`
- `np.load`
- `np.loadtxt`
