

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with median for numeric and mode for categorical

```python
# trying to fill missing values in my datset using for loop 
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])

```

```python
#check the Nulls 
missing_data = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Percentage (%)': df.isnull().sum() / len(df) * 100
})

missing_data[missing_data['Missing Count'] > 0]
```

```python

plt.figure(figsize=(10, 4))
sns.boxplot(x=df['age'], color='#00adb5')
plt.title('Age Distribution and Outliers Detection')
plt.xlabel('Age (Years)')
plt.show()

Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + 1.5 * IQR

iqr_outliers_count = (df['age'] > upper_limit).sum()
impossible_age_count = (df['age'] > 100).sum()

print(f"IQR Upper Limit: {upper_limit:.1f} years")
print(f"Number of statistical outliers (IQR): {iqr_outliers_count}")
print(f"Number of biologically impossible outliers (> 100): {impossible_age_count}")
```

# <a id="9">    9 ) . How long is the duration of the trip for each gender in minutes?  </a>

```python
# Make visualization to easy get insight

# ------------------------Duration distribution by member gender
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=df,
    x='member_gender',
    y='time_minutes',
    palette=['#3282b8', "#75120f"],
    errorbar=None,
)

for p in ax.patches:
    height = p.get_height()
    ax.annotate(
        f'{height:.1f} min',
        (p.get_x() + p.get_width() / 2.0, height / 2),
        ha='center',
        va='center',
        fontsize=11,
        color='white',
        weight='bold',
    )

plt.title('Average Trip Duration by Member Gender (in Minutes)', fontsize=13)
plt.xlabel('Member Gender', fontsize=11)
plt.ylabel('Average Duration (Minutes)', fontsize=11)
plt.show()
```

# <a id="1100"> Conclusions </a>


- Gender Distribution: Most users were Male (making up around 75% of total members).  

- Trip Duration Behavior: Most of the trips lasted 30 minutes or less, with average duration concentrated between 8 to 12 minutes as observed in both the minutes and hours distribution charts

- User Type & Trip Duration: Customers take significantly longer trips on average (~24 minutes) compared to Subscribers (~11 minutes).  

- User Sharing Behavior: The majority of users (over 90%) are not enrolled in the bike_share_for_all_trip program.  

- Age Distribution: User age is right-skewed, showing that the primary target audience and most active riders are concentrated between 25 and 40 years old.  

- Variable Correlations: There is no linear correlation between trip duration (duration_sec / time_minutes / time_Hour) and any  variables in the dataset.  

- Bike Utilization: There is no significant correlation between bike_id and any usage or demographic metrics, indicating that bikes are uniformly distributed across rides.  


## limitation 
- Missing Data: Missing entries were found in columns such as member_birth_year, member_gender, and station details, which had to be handled using median and mode imputation.   

- Age Outliers: Biological outliers were present in the age column with values exceeding 100 years (reaching up to 140+ years due to incorrect birth years like 1878), requiring filtering out values above 80 years. 

- Duration Outliers: Extreme duration values (reaching up to 1,400+ minutes) skewed standard visualizations, requiring log scaling and boxplot outlier suppression to clearly observe the under-30-minute trend.
